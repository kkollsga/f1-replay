# F1 Race Viewer - 2024 Monaco Grand Prix

Launch the interactive race viewer for the 2024 Monaco Grand Prix (Round 8)

```python
from f1_race import Manager

# Create manager
manager = Manager()

# Launch 2024 Monaco Grand Prix (Round 8) race viewer
# The viewer will start on http://localhost:5002
manager.race(2024, 8, port=5002, debug=False, force_update=False)
```
